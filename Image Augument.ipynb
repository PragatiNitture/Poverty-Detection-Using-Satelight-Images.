```python
import os
from keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
import numpy as np

# Define the path to the dataset
dataset_path = 'Data'

# Define a simpler set of augmentation techniques
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)
# Number of augmented images per original image
num_augmented_images = 3

# Iterate through each breed folder
for breed in os.listdir(dataset_path):
    breed_path = os.path.join(dataset_path, breed)
    train_path = os.path.join(breed_path, 'train')

    if os.path.isdir(train_path):
        print("came")
        # Iterate through each image in the train folder
        for img_name in os.listdir(train_path):
            img_path = os.path.join(train_path, img_name)
            if img_path.endswith('.jpg') or img_path.endswith('.png'):
                img = load_img(img_path)  # Load the image
                x = img_to_array(img)  # Convert the image to an array
                x = np.expand_dims(x, axis=0)  # Reshape the array
               
                # Generate augmented images
                i = 0
                for batch in datagen.flow(x, batch_size=1, save_to_dir=train_path, save_prefix='aug', save_format='jpg'):
                    
                    i += 1
                    if i >= num_augmented_images:  # Number of augmentations per image
                        
                        break
                        

print("Image augmentation completed.")
```
